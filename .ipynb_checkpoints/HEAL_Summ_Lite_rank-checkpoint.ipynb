# HEAL-Summ-Lite: Lightweight Ethical Health Summarization Pipeline## Architecture (5 Stages — Hybrid Extractive-Abstractive)```Original Text (600-700 words)  → [Stage 1] TextRank Key Sentence Extraction    (Python, instant)  → [Stage 2] One-Shot Phi-3 Mini Summarization    (1 LLM call, ~2-5 min)  → [Stage 3] Readability Scoring (FKGL + FRE)     (Python, instant)  → [Stage 4] 10 Rule-Based Heuristic Checks       (Python, instant)  → [Stage 5] Tiered Human Review Decision          (Python, instant)  → Results Tables + Severity Report```## Key Design Decisions- **TextRank extraction** solves sequential paraphrasing bias — ensures ALL sections are covered- **One-shot prompting** teaches the model what good summarization looks like- **Sentence-count prompt** (8-10 sentences) instead of word count — small LLMs can count sentences, not words- **`num_predict=240`** tokens ≈ 180 word ceiling — model wraps up naturally (NOT trimming)- **No trimming** — raw model output is preserved. Heuristics flag deviations honestly- **10 deterministic heuristics** — instant, reproducible, auditable quality checks

## Step 4: TextRank Key Sentence Extraction**Why TextRank?** The raw Phi-3 Mini suffers from **sequential paraphrasing bias** — it rephrases from the beginning of the text and never reaches later sections. TextRank solves this by selecting important sentences from ALL paragraphs before the LLM sees the text.**How TextRank Works:**1. Split text into sentences2. Compute similarity between every pair of sentences (shared words)3. Build a graph — each sentence is a node, similarity scores are edges4. Run PageRank algorithm (same one Google uses to rank web pages)5. Sentences connected to many other important sentences score highest6. Select top N sentences, preserving original order**Why TextRank over TF-IDF?** TextRank understands *relationships between sentences*, not just individual word importance. A safety caveat sentence that connects to treatment, diagnosis, and prevention topics gets boosted — even if its individual words aren't rare.

````python
def textrank_extract(text, num_sentences=10):    """    TextRank key sentence extraction (graph-based, inspired by Google PageRank).        Extracts the most important sentences from ALL sections of the document,    solving the sequential paraphrasing bias of small LLMs.        Algorithm:        1. Split into sentences        2. Compute pairwise sentence similarity (word overlap)        3. Build similarity matrix (graph adjacency matrix)        4. Run iterative PageRank until convergence        5. Return top N sentences in original order        No external dependencies — uses numpy only.    """    # Step 1: Split into sentences    sentences = re.split(r'(?<=[.!?])\s+', text.strip())    sentences = [s.strip() for s in sentences if len(s.split()) > 3]        if len(sentences) <= num_sentences:        return text, sentences  # Text is already short enough        # Step 2: Tokenize each sentence into words (lowercased, no punctuation)    def tokenize(sent):        words = re.findall(r'[a-z0-9]+', sent.lower())        # Remove common stop words that don't carry meaning        stop_words = {'the','a','an','is','are','was','were','be','been',                      'being','have','has','had','do','does','did','will',                      'would','could','should','may','might','shall','can',                      'to','of','in','for','on','with','at','by','from',                      'as','into','through','during','before','after',                      'and','but','or','nor','not','so','yet','both',                      'each','this','that','these','those','it','its'}        return [w for w in words if w not in stop_words and len(w) > 1]        tokenized = [set(tokenize(s)) for s in sentences]    n = len(sentences)        # Step 3: Build similarity matrix (word overlap / Jaccard-like)    sim_matrix = np.zeros((n, n))    for i in range(n):        for j in range(n):            if i != j and len(tokenized[i]) > 0 and len(tokenized[j]) > 0:                overlap = len(tokenized[i] & tokenized[j])                # Normalize by log of sentence lengths to avoid bias toward long sentences                norm = math.log(len(tokenized[i]) + 1) + math.log(len(tokenized[j]) + 1)                if norm > 0:                    sim_matrix[i][j] = overlap / norm        # Normalize columns (make it a proper transition matrix)    col_sums = sim_matrix.sum(axis=0)    col_sums[col_sums == 0] = 1  # Avoid division by zero    sim_matrix = sim_matrix / col_sums        # Step 4: Run PageRank iteration    damping = 0.85    scores = np.ones(n) / n  # Start with equal scores        for _ in range(50):  # 50 iterations is more than enough for convergence        new_scores = (1 - damping) / n + damping * sim_matrix.dot(scores)        if np.abs(new_scores - scores).sum() < 1e-6:  # Converged            break        scores = new_scores        # Step 5: Select top N sentences, preserve original order    ranked_indices = np.argsort(scores)[::-1][:num_sentences]    selected_indices = sorted(ranked_indices)  # Restore original order        selected_sentences = [sentences[i] for i in selected_indices]    extracted_text = " ".join(selected_sentences)        return extracted_text, selected_sentences# Test TextRank on first texttest_text = HEALTH_TEXTS[0]["text"]extracted, sents = textrank_extract(test_text, num_sentences=10)print(f"✓ TextRank ready.")print(f"  Test: {len(test_text.split())} words → {len(extracted.split())} words ({len(sents)} sentences)")print(f"  Compression: {len(extracted.split()) / len(test_text.split()):.0%}")
````

## Step 5: One-Shot Prompt & Ollama Interface**Why one-shot instead of zero-shot?**  Zero-shot: Model has NO example → defaults to sequential paraphrasing.  One-shot: Model sees ONE example of a good summary → mimics the pattern of covering ALL sections.**Why only one example?** Each example adds ~850 tokens. With `num_ctx=4096`, one example (850) + extracted text (~400) + system prompt (200) + output (240) = ~1700 tokens. Two examples would use 2550 tokens — still fits, but leaves less room. One example is the sweet spot for quality vs. space.

````python
# =============================================================================# ONE-SHOT EXAMPLE# =============================================================================# This example teaches the model: "cover ALL sections, not just the beginning"# The example input has: definition, statistics, types, complications, prevention# The example output covers ALL of these — not just the first 2-3 topics.# =============================================================================ONE_SHOT_EXAMPLE_INPUT = """Asthma is a chronic respiratory condition that affects the airways in the lungs. During an asthma attack, the airways become inflamed and narrowed, making it difficult to breathe. Common triggers include allergens, cold air, exercise, and respiratory infections. According to WHO, approximately 262 million people worldwide were affected by asthma in 2019, and it caused 455,000 deaths that year. Asthma is the most common chronic disease among children. While asthma cannot be cured, proper management with inhaled corticosteroids and bronchodilators can control symptoms effectively. Patients should avoid known triggers, follow their asthma action plan, and consult a healthcare professional for personalized treatment."""ONE_SHOT_EXAMPLE_OUTPUT = """Asthma is a chronic respiratory condition where airways become inflamed and narrowed during attacks, causing breathing difficulty. Common triggers include allergens, cold air, exercise, and respiratory infections that can worsen symptoms. The WHO reports that approximately 262 million people worldwide had asthma in 2019, causing 455,000 deaths. Asthma is the most common chronic disease among children globally. While no cure exists, inhaled corticosteroids and bronchodilators can effectively control asthma symptoms when used properly. Patients should avoid known triggers and follow their personalized asthma action plan consistently. Consult a healthcare professional before making medical decisions about asthma treatment."""SUMMARIZATION_SYSTEM_PROMPT = """You are a public health communication specialist. Your task is to summarize health information for the general public.RULES:1. Write EXACTLY 8 to 10 sentences. Each sentence should be 15 to 20 words long.2. Use plain language at an 8th-grade reading level.3. PRESERVE all statistics, percentages, and numbers exactly as stated.4. PRESERVE all named entities: disease names, drug names, organizations, countries.5. Cover ALL topics in the text — beginning, middle, AND end sections.6. If the text mentions treatments or medications, include: "Consult a healthcare professional before making medical decisions."7. Do NOT add information not in the original text.8. Do NOT round or simplify any numbers.9. Write in clear paragraphs. No bullet points or titles."""def build_summarization_prompt(extracted_text):    """    Builds a one-shot prompt with the extracted key sentences.    The example teaches the model to cover ALL sections, not just the beginning.    """    return f"""{SUMMARIZATION_SYSTEM_PROMPT}Here is an example:INPUT TEXT:{ONE_SHOT_EXAMPLE_INPUT}GOOD SUMMARY:{ONE_SHOT_EXAMPLE_OUTPUT}Now summarize the following health text the same way — cover ALL topics from beginning to end:INPUT TEXT:{extracted_text}GOOD SUMMARY:"""def call_ollama(prompt, temperature):    """    Sends prompt to Phi-3 Mini via Ollama.        Key settings:      num_predict=240: ~180 words max — model wraps up naturally (NOT trimming)      num_ctx=4096:    Fits one-shot example + extracted text + output      num_thread=4:    Half CPU, leaves room for Windows/OS    """    payload = {        "model": MODEL_NAME,        "prompt": prompt,        "stream": False,        "options": {            "temperature": temperature,            "top_p": 0.9,            "num_predict": 240,            "num_ctx": 4096,            "num_thread": 4,            "repeat_penalty": 1.1,        },    }    try:        response = requests.post(OLLAMA_URL, json=payload, timeout=600)        response.raise_for_status()    except requests.exceptions.ConnectionError:        print("\n❌ Cannot connect to Ollama! Make sure it is running:")        print("   ollama serve")        return None    except requests.exceptions.ReadTimeout:        print("\n❌ Timeout (>10 min). Close other apps to free RAM/CPU.")        return None    except requests.exceptions.HTTPError as e:        print(f"\n❌ Ollama error: {e}")        return None    result = response.json()    return result.get("response", "").strip() or Noneprint("✓ One-shot prompt and Ollama interface ready.")
````

## Step 6: Readability Scoring

````python
def compute_readability(text):    """FKGL = grade level (8 = 8th grader). FRE = reading ease (60-70 = standard)."""    return {        "fkgl": round(textstat.flesch_kincaid_grade(text), 1),        "fre": round(textstat.flesch_reading_ease(text), 1),    }print("✓ Readability scoring ready.")
````

## Step 7: 10 Rule-Based Heuristic Checks| # | Check | Catches | Severity ||---|---|---|---|| 1 | Numeric Coverage | Missing statistics | WARNING || 2 | Entity Coverage | Missing drug/org names | WARNING || 3 | Compression Ratio | Barely shorter or too aggressive | WARNING || 4 | Caveat Presence | Treatment without safety disclaimer | CRITICAL || 5 | Negation Flip | "does NOT spread" → "spreads" | CRITICAL || 6 | Hedging Preservation | "may cause" → "causes" | WARNING || 7 | Source Attribution | Missing WHO/CDC/NHS | INFO || 8 | Sentence Count | Too few or too many | WARNING || 9 | Semantic Similarity | Hallucination or copy-paste | CRITICAL || 10 | Sensitive Topics | Missing crisis helpline | CRITICAL |

````python
# --- HELPER: Extract numbers from text ---def extract_numbers(text):    """Extracts all numeric values including percentages, decimals, comma-separated."""    raw = re.findall(r'\d[\d,]*\.?\d*%?', text)    return set(num.replace(",", "") for num in raw)# --- HELPER: Extract key entities ---def extract_key_entities(text):    """Extracts organization names, drug names, and disease terms."""    entities = set()    # Health organizations    orgs = [        "WHO", "CDC", "NHS", "World Health Organization",        "Centers for Disease Control", "National Health Service",        "American Psychological Association", "National Institute of Mental Health",        "National Sleep Foundation",    ]    for org in orgs:        if org.lower() in text.lower():            entities.add(org)    # Drug and treatment patterns    drugs = re.findall(        r'\b(ACE inhibitors?|ARBs?|calcium channel blockers?|diuretics?|'        r'beta blockers?|antiviral drugs?|RTS,S/AS01|Mosquirix|R21/Matrix-M|'        r'insecticide-treated nets?|ITNs?|IRS|insulin|chemoprevention|'        r'oseltamivir|Tamiflu|zanamivir|Relenza|peramivir|Rapivab|'        r'baloxavir marboxil|Xofluza|artemisinin|ACT|'        r'enalapril|lisinopril|amlodipine|CBT|'        r'cognitive behavioral therapy)\b',        text, re.IGNORECASE    )    for drug in drugs:        entities.add(drug.lower())    # Disease names    diseases = re.findall(        r'\b(diabetes|malaria|influenza|flu|hypertension|'        r'high blood pressure|pneumonia|sepsis|HIV|AIDS|'        r'obesity|retinopathy|neuropathy|'        r'anxiety|depression|stroke|heart attack)\b',        text, re.IGNORECASE    )    for disease in diseases:        entities.add(disease.lower())    return entities# ──────────────────────────────────────────────────────────────────────# CHECK 1: Numeric Coverage# ──────────────────────────────────────────────────────────────────────def check_numeric_coverage(original, summary):    """What % of numbers from the original appear in the summary?"""    orig_nums = extract_numbers(original)    summ_nums = extract_numbers(summary)    if not orig_nums:        return {"status": "PASS", "severity": "INFO",                "reason": "No numbers in original", "detail": {}}    missing = orig_nums - summ_nums    coverage = round((len(orig_nums) - len(missing)) / len(orig_nums) * 100, 1)    if coverage < 30:        return {"status": "FLAG", "severity": "WARNING",                "reason": f"Only {coverage}% of numbers preserved ({len(missing)} missing)",                "detail": {"coverage": coverage, "missing": list(missing)[:5]}}    return {"status": "PASS", "severity": "INFO",            "reason": f"{coverage}% of numbers preserved",            "detail": {"coverage": coverage, "missing": list(missing)[:5]}}# ──────────────────────────────────────────────────────────────────────# CHECK 2: Entity Coverage# ──────────────────────────────────────────────────────────────────────def check_entity_coverage(original, summary):    """What % of key entities from the original appear in the summary?"""    orig_ents = extract_key_entities(original)    summ_ents = extract_key_entities(summary)    if not orig_ents:        return {"status": "PASS", "severity": "INFO",                "reason": "No key entities detected", "detail": {}}    orig_lower = {e.lower() for e in orig_ents}    summ_lower = {e.lower() for e in summ_ents}    missing = orig_lower - summ_lower    coverage = round((len(orig_lower) - len(missing)) / len(orig_lower) * 100, 1)    if coverage < 40:        return {"status": "FLAG", "severity": "WARNING",                "reason": f"Only {coverage}% of entities preserved ({len(missing)} missing)",                "detail": {"coverage": coverage, "missing": list(missing)[:5]}}    return {"status": "PASS", "severity": "INFO",            "reason": f"{coverage}% of entities preserved",            "detail": {"coverage": coverage, "missing": list(missing)[:5]}}# ──────────────────────────────────────────────────────────────────────# CHECK 3: Compression Ratio# ──────────────────────────────────────────────────────────────────────def check_compression_ratio(original, summary):    """Is the summary appropriately shorter than the original?"""    orig_words = len(original.split())    summ_words = len(summary.split())    ratio = round(summ_words / orig_words, 2) if orig_words > 0 else 0    if ratio > 0.85:        return {"status": "FLAG", "severity": "WARNING",                "reason": f"Ratio {ratio} — summary barely shorter than original",                "detail": {"ratio": ratio}}    if ratio < 0.10:        return {"status": "FLAG", "severity": "WARNING",                "reason": f"Ratio {ratio} — summary may be too aggressive",                "detail": {"ratio": ratio}}    return {"status": "PASS", "severity": "INFO",            "reason": f"Compression ratio {ratio} — within acceptable range",            "detail": {"ratio": ratio}}# ──────────────────────────────────────────────────────────────────────# CHECK 4: Caveat / Safety Language Check# ──────────────────────────────────────────────────────────────────────def check_caveat_presence(original, summary):    """If original discusses treatments, does summary have safety language?"""    treatment_keywords = [        "treatment", "treated", "medication", "medicine", "prescribe",        "drug", "dosage", "dose", "vaccine", "therapy", "antiviral",        "inhibitor", "surgery", "insulin", "chemoprevention",    ]    caveat_phrases = [        "consult", "healthcare professional", "medical professional",        "doctor", "physician", "seek medical", "professional advice",        "health provider", "medical advice", "medical decisions",    ]    has_treatment = any(kw in original.lower() for kw in treatment_keywords)    has_caveat = any(phrase in summary.lower() for phrase in caveat_phrases)    if has_treatment and not has_caveat:        return {"status": "FLAG", "severity": "CRITICAL",                "reason": "Treatment content found but NO safety caveat in summary",                "detail": {"has_treatment": True, "has_caveat": False}}    if has_treatment and has_caveat:        return {"status": "PASS", "severity": "INFO",                "reason": "Treatment content found; safety caveat present",                "detail": {"has_treatment": True, "has_caveat": True}}    return {"status": "PASS", "severity": "INFO",            "reason": "No treatment content — caveat not required",            "detail": {"has_treatment": False, "has_caveat": has_caveat}}# ──────────────────────────────────────────────────────────────────────# CHECK 5: Negation Flip Detection  [NEW]# ──────────────────────────────────────────────────────────────────────# Why this matters: "does NOT spread from person to person" becoming# "spreads from person to person" is the most dangerous error possible# in health summarization. One missing "not" reverses the entire meaning.# ──────────────────────────────────────────────────────────────────────def check_negation_flip(original, summary):    """Detects if critical negations from the original are missing in summary."""    # Extract negation patterns from original    negation_patterns = [        r"(not|no|never|neither|nor|cannot|can't|don't|doesn't|didn't|won't|shouldn't|isn't|aren't)\s+(\w+(?:\s+\w+){0,4})",    ]    original_negations = []    for pattern in negation_patterns:        matches = re.findall(pattern, original.lower())        for match in matches:            neg_word, context = match            original_negations.append((neg_word, context.strip()))    if not original_negations:        return {"status": "PASS", "severity": "INFO",                "reason": "No critical negations found in original",                "detail": {}}    # Check if the context phrases appear in summary WITHOUT their negation    potential_flips = []    for neg_word, context in original_negations:        # Skip very short/common contexts        if len(context.split()) < 2:            continue        # Check if context appears in summary        if context in summary.lower():            # Context is there — check if negation is also there            # Look in a window around the context            context_pos = summary.lower().find(context)            window_start = max(0, context_pos - 30)            window = summary.lower()[window_start:context_pos + len(context) + 10]            neg_words = ["not", "no", "never", "cannot", "can't", "don't",                         "doesn't", "didn't", "won't", "shouldn't", "isn't", "aren't"]            if not any(nw in window for nw in neg_words):                potential_flips.append(f"'{neg_word} {context}' may be flipped")    if potential_flips:        return {"status": "FLAG", "severity": "CRITICAL",                "reason": f"Possible negation flip: {potential_flips[0]}",                "detail": {"flips": potential_flips}}    return {"status": "PASS", "severity": "INFO",            "reason": "No negation flips detected",            "detail": {"checked": len(original_negations)}}# ──────────────────────────────────────────────────────────────────────# CHECK 6: Hedging Language Preservation  [NEW]# ──────────────────────────────────────────────────────────────────────# Why: "may contribute to obesity" vs "causes obesity" — removing# uncertainty language makes speculative claims sound definitive.# ──────────────────────────────────────────────────────────────────────def check_hedging_preservation(original, summary):    """Checks if uncertain/speculative language is preserved."""    hedging_phrases = [        "may", "might", "could", "possibly", "potentially",        "suggests", "preliminary", "estimated", "approximately",        "believed to", "thought to", "appears to", "seems to",        "is associated with", "may contribute", "more research",    ]    orig_hedges = [h for h in hedging_phrases if h in original.lower()]    summ_hedges = [h for h in hedging_phrases if h in summary.lower()]    if not orig_hedges:        return {"status": "PASS", "severity": "INFO",                "reason": "No hedging language in original",                "detail": {}}    # Check if hedging topics are in summary but WITHOUT hedging    missing_hedges = [h for h in orig_hedges if h not in summary.lower()]    preservation = round((len(orig_hedges) - len(missing_hedges)) / len(orig_hedges) * 100, 1)    if preservation < 30 and len(orig_hedges) >= 2:        return {"status": "FLAG", "severity": "WARNING",                "reason": f"Only {preservation}% hedging language preserved — claims may sound more definitive than intended",                "detail": {"original_hedges": orig_hedges, "missing": missing_hedges}}    return {"status": "PASS", "severity": "INFO",            "reason": f"{preservation}% hedging language preserved",            "detail": {"original_hedges": orig_hedges, "preserved": len(orig_hedges) - len(missing_hedges)}}# ──────────────────────────────────────────────────────────────────────# CHECK 7: Source Attribution  [NEW]# ──────────────────────────────────────────────────────────────────────# Why: Health summaries without source attribution lose credibility.# A summary saying "1.5 million deaths" without saying "according to WHO"# looks like an unsourced claim.# ──────────────────────────────────────────────────────────────────────def check_source_attribution(original, summary, source_name):    """Does the summary mention the source organization?"""    # Build list of source identifiers    source_terms = []    if "WHO" in source_name or "World Health" in source_name:        source_terms = ["who", "world health organization"]    elif "CDC" in source_name or "Centers for Disease" in source_name:        source_terms = ["cdc", "centers for disease control"]    elif "NHS" in source_name or "National Health Service" in source_name:        source_terms = ["nhs", "national health service"]    if not source_terms:        return {"status": "PASS", "severity": "INFO",                "reason": "Source type not tracked", "detail": {}}    has_attribution = any(term in summary.lower() for term in source_terms)    if not has_attribution:        return {"status": "FLAG", "severity": "INFO",                "reason": f"Summary does not mention source ({source_name})",                "detail": {"source": source_name}}    return {"status": "PASS", "severity": "INFO",            "reason": f"Source attribution present ({source_name})",            "detail": {"source": source_name}}# ──────────────────────────────────────────────────────────────────────# CHECK 8: Sentence Count Check  [NEW]# ──────────────────────────────────────────────────────────────────────# Why: Too few sentences = probably copied a chunk verbatim.# Too many sentences = not really summarized, just shortened each one.# ──────────────────────────────────────────────────────────────────────def check_sentence_count(summary):    """Verifies summary has a reasonable number of sentences."""    sentences = re.split(r'(?<=[.!?])\s+', summary.strip())    count = len(sentences)    if count < 3:        return {"status": "FLAG", "severity": "WARNING",                "reason": f"Only {count} sentences — summary may lack detail",                "detail": {"sentence_count": count}}    if count > 12:        return {"status": "FLAG", "severity": "WARNING",                "reason": f"{count} sentences — summary may not be condensed enough",                "detail": {"sentence_count": count}}    return {"status": "PASS", "severity": "INFO",            "reason": f"{count} sentences — appropriate for summary length",            "detail": {"sentence_count": count}}# ──────────────────────────────────────────────────────────────────────# CHECK 9: Semantic Similarity Score  [NEW]# ──────────────────────────────────────────────────────────────────────# Why: A summary that drifts off-topic or hallucinates will have low# textual overlap with the original. Uses SequenceMatcher for a# lightweight similarity score without any ML dependencies.# ──────────────────────────────────────────────────────────────────────def check_semantic_similarity(original, summary):    """Computes text similarity between original and summary."""    # Normalize both texts    orig_clean = re.sub(r'\s+', ' ', original.lower().strip())    summ_clean = re.sub(r'\s+', ' ', summary.lower().strip())    # Use difflib SequenceMatcher for lightweight similarity    similarity = round(difflib.SequenceMatcher(None, orig_clean, summ_clean).ratio() * 100, 1)    if similarity < 15:        return {"status": "FLAG", "severity": "CRITICAL",                "reason": f"Similarity {similarity}% — summary may contain hallucinated content",                "detail": {"similarity_pct": similarity}}    if similarity > 85:        return {"status": "FLAG", "severity": "WARNING",                "reason": f"Similarity {similarity}% — summary may be too close to copy-paste",                "detail": {"similarity_pct": similarity}}    return {"status": "PASS", "severity": "INFO",            "reason": f"Similarity {similarity}% — good balance of faithfulness and compression",            "detail": {"similarity_pct": similarity}}# ──────────────────────────────────────────────────────────────────────# CHECK 10: Sensitive Topic Flag  [NEW]# ──────────────────────────────────────────────────────────────────────# Why: Topics like suicide, self-harm, addiction, mental health crisis# require extra care. If the original mentions helpline numbers or# crisis resources, the summary MUST preserve them.# ──────────────────────────────────────────────────────────────────────def check_sensitive_topics(original, summary):    """Flags if sensitive health topics need special handling."""    sensitive_keywords = [        "suicide", "self-harm", "self harm", "crisis", "addiction",        "overdose", "eating disorder", "substance abuse", "mental health crisis",        "domestic violence", "abuse",    ]    resource_patterns = [        r"988", r"911", r"helpline", r"hotline", r"crisis line",        r"lifeline", r"emergency", r"call or text",    ]    has_sensitive = any(kw in original.lower() for kw in sensitive_keywords)    if not has_sensitive:        return {"status": "PASS", "severity": "INFO",                "reason": "No sensitive topics detected",                "detail": {}}    # Check if crisis resources from original are in summary    orig_resources = [r for r in resource_patterns if re.search(r, original.lower())]    summ_resources = [r for r in resource_patterns if re.search(r, summary.lower())]    missing_resources = set(orig_resources) - set(summ_resources)    if missing_resources:        return {"status": "FLAG", "severity": "CRITICAL",                "reason": f"Sensitive topic detected but crisis resources missing from summary",                "detail": {"missing_resources": list(missing_resources)}}    return {"status": "PASS", "severity": "INFO",            "reason": "Sensitive topic detected; crisis resources preserved",            "detail": {"resources_found": summ_resources}}# =============================================================================# RUN ALL 10 HEURISTICS# =============================================================================def run_all_heuristics(original, summary, source_name):    """    Runs all 10 rule-based heuristic checks and returns results dict.    All checks are deterministic Python — no LLM calls needed.    """    return {        "numeric_coverage":     check_numeric_coverage(original, summary),        "entity_coverage":      check_entity_coverage(original, summary),        "compression_ratio":    check_compression_ratio(original, summary),        "caveat_presence":      check_caveat_presence(original, summary),        "negation_flip":        check_negation_flip(original, summary),        "hedging_preservation": check_hedging_preservation(original, summary),        "source_attribution":   check_source_attribution(original, summary, source_name),        "sentence_count":       check_sentence_count(summary),        "semantic_similarity":  check_semantic_similarity(original, summary),        "sensitive_topics":     check_sensitive_topics(original, summary),    }print("✓ All 10 heuristic checks ready.")
````

## Step 8: Tiered Human Review| Tier | Condition | Action ||---|---|---|| **CRITICAL** | Any CRITICAL flag | Always escalate || **ESCALATE** | 2+ WARNING flags | Escalate || **REVIEW** | 1 WARNING flag | Soft flag || **APPROVED** | No flags | Safe to publish |

````python
def decide_human_review(word_count, readability, heuristics):    """    Determines escalation level based on severity tiers.    Returns:        Tuple of (decision: str, reasons: list[str])        decision is one of: "CRITICAL", "ESCALATE", "APPROVED"    """    critical_reasons = []    warning_reasons = []    info_notes = []    # Analyze heuristic flags by severity    for check_name, result in heuristics.items():        if result["status"] == "FLAG":            severity = result.get("severity", "WARNING")            tag = f"[{severity}] {check_name}: {result['reason']}"            if severity == "CRITICAL":                critical_reasons.append(tag)            elif severity == "WARNING":                warning_reasons.append(tag)            else:                info_notes.append(tag)    # Readability checks (WARNING level)    if readability["fkgl"] > MAX_FKGL_GRADE:        warning_reasons.append(            f"[WARNING] readability: FKGL grade {readability['fkgl']} exceeds {MAX_FKGL_GRADE}")    if readability["fre"] < FRE_MINIMUM:        warning_reasons.append(            f"[WARNING] readability: FRE {readability['fre']} below {FRE_MINIMUM}")    # Word count check (WARNING level)    if word_count < MIN_WORD_COUNT:        warning_reasons.append(            f"[WARNING] word_count: {word_count} below minimum {MIN_WORD_COUNT}")    elif word_count > MAX_WORD_COUNT:        warning_reasons.append(            f"[WARNING] word_count: {word_count} exceeds maximum {MAX_WORD_COUNT}")    # Decision logic    all_reasons = critical_reasons + warning_reasons + info_notes    if critical_reasons:        return "CRITICAL", all_reasons    elif len(warning_reasons) >= 2:        return "ESCALATE", all_reasons    elif len(warning_reasons) == 1:        return "REVIEW", all_reasons  # Soft flag — one warning, not urgent    else:        return "APPROVED", info_notesprint("✓ Tiered human review ready.")
````

## Step 9: Main Pipeline (5 Stages)```[1/5] TextRank Extraction       → instant (Python)[2/5] One-Shot Summarization    → ~2-5 min (Phi-3 Mini, single LLM call)[3/5] Readability Scoring       → instant (textstat)[4/5] 10 Heuristic Checks      → instant (Python)[5/5] Tiered Human Review       → instant (Python)```

````python
def process_single_text(health_text):    """    Runs the full HEAL-Summ-Lite pipeline on one health text.        5 Stages:        1. TextRank key sentence extraction (covers ALL sections)        2. One-shot prompted summarization via Phi-3 Mini        3. Readability scoring (FKGL + FRE)        4. 10 rule-based heuristic checks (deterministic)        5. Tiered human review decision (severity-based)        No trimming — raw model output preserved as-is.    """    text_id = health_text["id"]    source = health_text["source"]    title = health_text["title"]    original = health_text["text"]    print(f"\n{'='*70}")    print(f"  Processing: {text_id} — {title}")    print(f"  Source: {source}")    print(f"  Original: {len(original.split())} words")    print(f"{'='*70}")    # --- Stage 1: TextRank Extraction ---    print("\n  [1/5] TextRank key sentence extraction...")    extracted_text, selected_sents = textrank_extract(original, num_sentences=10)    ext_words = len(extracted_text.split())    print(f"        Extracted {len(selected_sents)} key sentences ({ext_words} words)")    print(f"        Compression: {len(original.split())} → {ext_words} words")    # --- Stage 2: One-Shot Summarization ---    print(f"  [2/5] One-shot summarization (Phi-3 Mini, temp={SUMMARIZATION_TEMPERATURE})...")    print("        This may take 2-5 minutes on CPU. Please wait...")    start_time = time.time()    prompt = build_summarization_prompt(extracted_text)    summary = call_ollama(prompt, temperature=SUMMARIZATION_TEMPERATURE)    gen_time = round(time.time() - start_time, 1)    if summary is None:        print("        ❌ Summarization failed. Check Ollama.")        return None    word_count = len(summary.split())    print(f"        ✓ Done in {gen_time}s — {word_count} words (raw, no trimming)")    # --- Stage 3: Readability ---    print("  [3/5] Computing readability metrics...")    readability = compute_readability(summary)    print(f"        FKGL: {readability['fkgl']} (grade level) | FRE: {readability['fre']} (ease)")    # --- Stage 4: 10 Heuristic Checks ---    print("  [4/5] Running 10 rule-based heuristic checks...")    heuristics = run_all_heuristics(original, summary, source)    flags = sum(1 for v in heuristics.values() if v["status"] == "FLAG")    passes = sum(1 for v in heuristics.values() if v["status"] == "PASS")    print(f"        Results: {passes} PASS, {flags} FLAG")    for name, result in heuristics.items():        marker = "✓" if result["status"] == "PASS" else "⚠"        sev = result.get("severity", "")        print(f"          {marker} [{sev:8s}] {name}: {result['reason']}")    # --- Stage 5: Tiered Human Review ---    print("  [5/5] Applying tiered human review rules...")    decision, review_reasons = decide_human_review(word_count, readability, heuristics)    print(f"        Decision: {decision}")    for reason in review_reasons:        print(f"          ⚠ {reason}")    return {        "id": text_id, "source": source, "title": title,        "original_word_count": len(original.split()),        "extracted_word_count": ext_words,        "extracted_sentences": len(selected_sents),        "summary": summary, "summary_word_count": word_count,        "fkgl": readability["fkgl"], "fre": readability["fre"],        "heuristics": heuristics, "decision": decision,        "review_reasons": review_reasons, "generation_time": gen_time,    }print("✓ Pipeline function ready.")
````

````python
def display_results(results):    results = [r for r in results if r is not None]    if not results:        print("No results to display.")        return    print("\n\n" + "=" * 110)    print("  HEAL-Summ-Lite — RESULTS")    print("=" * 110)    # Table 1: Overview    print("\n  [TABLE 1] Summary Overview & Readability")    print("  " + "-" * 90)    t1 = [[r["id"],           r["title"][:22]+".." if len(r["title"])>22 else r["title"],           r["original_word_count"],           r["extracted_word_count"],           r["summary_word_count"],           r["fkgl"], r["fre"],           f"{r['generation_time']}s",           r["decision"]]          for r in results]    print(tabulate(t1, headers=["ID","Title","Orig\nWords","Extracted\nWords",                                 "Summary\nWords","FKGL","FRE","Time","Decision"],                   tablefmt="grid", stralign="center"))    # Table 2: Heuristics    print("\n  [TABLE 2] Heuristic Quality Checks (10 Checks)")    print("  " + "-" * 90)    check_names = ["numeric_coverage","entity_coverage","compression_ratio","caveat_presence",                   "negation_flip","hedging_preservation","source_attribution","sentence_count",                   "semantic_similarity","sensitive_topics"]    t2 = [[r["id"]] + [r["heuristics"][c]["status"] for c in check_names] for r in results]    print(tabulate(t2, headers=["ID","Num\nCov","Ent\nCov","Comp\nRat","Caveat",                                 "Neg\nFlip","Hedge","Src\nAttr","Sent\nCnt",                                 "Sem\nSim","Sens\nTop"],                   tablefmt="grid", stralign="center"))    # Flagged    flagged = [r for r in results if r["decision"] in ("CRITICAL","ESCALATE")]    if flagged:        print(f"\n  ⚠ FLAGGED FOR HUMAN REVIEW: {len(flagged)} of {len(results)}")        for r in flagged:            print(f"\n    [{r['decision']}] {r['id']} — {r['title']}")            for reason in r["review_reasons"]:                print(f"      ⚠ {reason}")    else:        print("\n  ✓ All summaries passed quality checks.")    # Summaries    print("\n\n" + "=" * 110)    print("  GENERATED SUMMARIES")    print("=" * 110)    for r in results:        print(f"\n  --- {r['id']}: {r['title']} ---")        print(f"  [{r['summary_word_count']} words | FKGL: {r['fkgl']} | FRE: {r['fre']} | {r['decision']}]")        print(f"\n  {r['summary']}")    # Detailed analysis    print("\n\n" + "=" * 110)    print("  DETAILED HEURISTIC ANALYSIS")    print("=" * 110)    for r in results:        print(f"\n  --- {r['id']}: {r['title']} ---")        for name, result in r["heuristics"].items():            marker = "✓" if result["status"] == "PASS" else "⚠"            sev = result.get("severity", "INFO")            print(f"    {marker} [{sev:8s}] {name}: {result['reason']}")            d = result.get("detail", {})            if d.get("missing"): print(f"               Missing: {d['missing']}")            if d.get("flips"): print(f"               Flips: {d['flips']}")def display_methodology():    print("\n\n" + "=" * 110)    print("  METHODOLOGY & TRANSPARENCY")    print("=" * 110)    print("""  Model:        Phi-3 Mini 3.8B (Microsoft) via Ollama — local CPU inference  LLM Usage:    Summarization ONLY (1 LLM call per text, no LLM evaluation)  Extraction:   TextRank (graph-based, PageRank-inspired sentence ranking)  Prompting:    One-shot — example teaches model to cover ALL sections  Readability:  textstat library (FKGL + FRE)  Word Count:   NOT enforced by trimming. Three strategies combined:                (a) TextRank pre-selects key sentences from ALL paragraphs                (b) Prompt uses sentence count (8-10 sentences × 15-20 words)                (c) num_predict=240 tokens ≈ 180 word natural generation ceiling                Raw model output is preserved as-is. Heuristics flag deviations.  Pipeline (5 Stages):    1. TextRank Extraction   — Graph-based key sentence selection (instant)    2. LLM Summarization     — One-shot prompted Phi-3 Mini (2-5 min)    3. Readability Scoring   — FKGL + FRE via textstat (instant)    4. 10 Heuristic Checks   — Deterministic Python checks (instant)    5. Tiered Human Review   — Severity-based escalation (instant)  Heuristic Checks (10):    1. Numeric Coverage       6. Hedging Preservation    2. Entity Coverage        7. Source Attribution    3. Compression Ratio      8. Sentence Count    4. Caveat Presence        9. Semantic Similarity    5. Negation Flip         10. Sensitive Topic Flag  Human Review Tiers:    CRITICAL  — Always escalate (negation flip, hallucination, missing crisis info)    ESCALATE  — 2+ warnings    REVIEW    — 1 warning (soft flag)    APPROVED  — All checks passed  Known Limitations:    1. Phi-3 Mini (3.8B) has limited abstractive capacity — TextRank compensates    2. TextRank uses word overlap, not semantic embeddings    3. Negation detection uses regex — may miss complex negations    4. One-shot example may bias toward its specific format    5. This is a first-pass screening tool, NOT a replacement for expert review""")print("✓ Display functions ready.")
````
